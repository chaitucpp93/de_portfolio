```python
import pandas as pd

#Reading the file

def read_files(filepath , encoding = 'utf-8', nrows = None ) :
    df = pd.read_csv(filepath, encoding = encoding , nrows = nrows)
    return df

#Cleaning the file

def clean_file(df, fill_value = 0 , drop_col = None) :
    # Step 1 — fill nulls
    df = df.fillna(fill_value)
    # Step 2 - Drop columns based the the drop_col
    if drop_col :
        df = df.drop(columns = drop_col)
    # Step 3 - reset index - safety
    df = df.reset_index(drop = True)
    return df

# Test it
df = read_files('sales_data.csv')
print("Before:", df.isnull().sum().sum(), "nulls")

df_clean = clean_file(df)
print("After:", df_clean.isnull().sum().sum(), "nulls")

df_clean2 = clean_file(df, fill_value=0, drop_col='discount')
print("After with drop:", df_clean2.columns.tolist())


```

```python
def transform_data(df,group_col,value_col,agg_func = 'sum') :
    
    result = df.groupby(group_col)[value_col].agg(agg_func).reset_index()

    # rename the value column to reflect which aggregration applied. Example sum_sales or mean_sales etc

    result.rename(columns={value_col: f'{agg_func}_{value_col}'}, inplace =True)

    return result
    
df = read_files('sales_data.csv')
city_sales = transform_data(df, 'city', 'sales')
print(city_sales) 

region_profit = transform_data(df, 'region', 'profit', agg_func='mean')
print(region_profit)

cat_orders = transform_data(df, 'category', 'sales', agg_func='count')
print(cat_orders)
```

```python
def run_pipeline(filepath, group_col, value_col, 
                 agg_func='sum', fill_value=0, drop_col=None):
    
    # Step 1 — Read
    df = read_files(filepath)
    print(f"Step 1 — Read: {df.shape}")
    
    # Step 2 — Clean
    df = clean_file(df, fill_value=fill_value, drop_col=drop_col)
    print(f"Step 2 — Clean: {df.isnull().sum().sum()} nulls remaining")
    
    # Step 3 — Transform
    result = transform_data(df, group_col, value_col, agg_func)
    print(f"Step 3 — Transform: {result.shape}")
    
    return result

# Run the full pipeline in one call
output = run_pipeline(
    filepath  = 'sales_data.csv',
    group_col = 'region',
    value_col = 'sales',
    agg_func  = 'sum',
    drop_col  = 'discount'
)

print(output)
```

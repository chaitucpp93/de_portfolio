```python
import pandas as pd
import json
import os

# Check your current working directory
print(os.getcwd())
```

```python
#create the sample data

data = {
    'city' : ['Mumbai','Delhi','Pune','Hyderabad'] ,
    'sales' : [100000, 200000, 90000,150000] ,
    'region' : ['West','North','West', 'South'] ,
    'active' : [True , True, False, True]
    
}

df = pd.DataFrame(data)

#write to csv

df.to_csv('sales_data.csv',index = False)
print("CSV written")

# write to json
df.to_json('sales_data.json', orient = 'records' , indent =2)
print('JSON written')

#write to excel
df.to_excel('sales_data.xlsx', index = False , sheet_name = 'Sales')
print('Excel written')
```

```python
# 1. Read only specific columns — don't load what you don't need
df_subset = pd.read_csv('sales_data.csv', usecols=['city', 'sales'])
print(df_subset)

# 2. Skip rows — some CSV files have metadata at the top
df_skip = pd.read_csv('sales_data.csv', skiprows=2)
print(df_skip)

# 3. Read only first N rows — useful for previewing large files
df_head = pd.read_csv('sales_data.csv', nrows=2)
print(df_head)

# 4. Handle missing values — custom null markers
df_na = pd.read_csv('sales_data.csv', na_values=['NA', 'N/A', 'missing', '-'])
print(df_na)
```

```python
# JSON edge case 1 — nested JSON (common in API responses)
nested = {
    'status': 'success',
    'data': [
        {'city': 'Mumbai', 'sales': 100000},
        {'city': 'Delhi',  'sales': 200000}
    ]
}

# Writing nested JSON
with open('nested.json', 'w') as f:
    json.dump(nested, f, indent=2)

# Reading nested JSON — pd.read_json won't work directly
with open('nested.json', 'r') as f:
    raw = json.load(f)

# Extract the data key first
df_nested = pd.DataFrame(raw['data'])
print(df_nested)
```

```python
# Create sample multi-level Excel file first
import pandas as pd
from itertools import product

# Build multi-level columns
arrays = [
    ['Q1', 'Q1', 'Q1', 'Q2', 'Q2', 'Q2'],
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
]
columns = pd.MultiIndex.from_arrays(arrays)

data = [
    [100, 120, 90,  110, 130, 150],
    [200, 180, 220, 190, 210, 240],
    [80,  90,  70,  85,  95,  100]
]

df_multi = pd.DataFrame(data, columns=columns, index=['Mumbai', 'Delhi', 'Pune'])
print(df_multi)

# Write to Excel
df_multi.to_excel('multi_level.xlsx', sheet_name='Quarterly')
```

```python
# Read without handling multi-level
df_read = pd.read_excel('multi_level.xlsx', sheet_name='Quarterly')
print(df_read)
print(df_read.columns.tolist())
```

```python
# Tell pandas — first TWO rows are both headers
df_read = pd.read_excel('multi_level.xlsx',
                        sheet_name='Quarterly',
                        index_col=0,      # first column is the index (city)
                        header=[0, 1])    # rows 0 and 1 are both headers

print(df_read)
print(df_read.columns.tolist())
```

```python
# Combine levels into one string — Q1_Jan, Q1_Feb etc
df_read.columns = ['_'.join(col) for col in df_read.columns]
print(df_read.columns.tolist())
# ['Q1_Jan', 'Q1_Feb', 'Q1_Mar', 'Q2_Apr', 'Q2_May', 'Q2_Jun']

print(df_read)
```

```python
# Reset index first — city is currently the index
df_reset = df_read.reset_index()

# Melt to long format
df_long = pd.melt(
    df_reset,
    id_vars    = 'city',
    value_vars = ['Q1_Jan', 'Q1_Feb', 'Q1_Mar', 'Q2_Apr', 'Q2_May', 'Q2_Jun'],
    var_name   = 'quarter_month',
    value_name = 'sales'
)

print(df_long)
```
